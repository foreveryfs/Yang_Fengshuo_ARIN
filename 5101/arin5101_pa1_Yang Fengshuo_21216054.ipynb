#### Question 2 (10 points)
**Task**: Given three lists from the Palmer Penguins dataset `islands`, `body_masses`, and `years` create a dictionary where keys are `(island, year)` tuples and values are the average body mass for penguins in that group.

**Instructions**:
- Use the three given lists: `islands`, `body_masses`, and `years` as a test.
- Create an empty dictionary `{ }` to collect body masses for each `(island, year)` group, i.e. using `(island, year)` as the dictionary key and body masses as the dictionary value.
- Use the `zip()` function to iterate over the lists and group body masses by `(island, year)` tuples, i.e. storing the body masses in a list.
- Use dictionary comprehension to compute the average body mass for each group. Hint: You can use `sum()` and `len()` to help you compute the average.
- Store the result in a variable `group_averages`.
- Print `group_averages` to verify that the program output matches the expected output.

**Sample Input**:
```python
islands = ['Biscoe', 'Biscoe', 'Torgersen', 'Torgersen']
body_masses = [3750, 3800, 3700, 3750]
years = [2007, 2007, 2008, 2008]
```
**Expected Output** (averages):
```python
{('Biscoe', 2007): 3775.0, ('Torgersen', 2008): 3725.0}
```

**Requirements**:
- You **must not** use any libraries to complete this question.
- You **must** use dictionary comprehension to complete this question.

````python
# These are the example input lists (we may use other lists when we grade your work)
islands = ['Biscoe', 'Biscoe', 'Torgersen', 'Torgersen']
body_masses = [3750, 3800, 3700, 3750]
years = [2007, 2007, 2008, 2008]

##### Don't change the above code

# Write your code here

# Create a empty dictionary to store body masses by (island, year)
group_masses_dic = {}

# Group body masses by (island, year) tuples
for disland,dmass,dyear in zip(islands, body_masses, years): #New index:dic+ origin
    key = (disland,dyear)
    if key not in group_masses_dic:      # Init
        group_masses_dic[key] = []
    group_masses_dic[key].append(dmass)  # Store

# Compute averages for each group
group_averages = {key: sum(cmass)/len(cmass) for key, cmass in group_masses_dic.items()} # Use dictionary comprehension here

##### Don't change the following code

# Print result
print(group_averages)
````

````python
# Load dataset
penguins = pd.read_csv("penguins.csv")

##### Don't change the above code

# Write your code here

# Drop rows with missing values in required columns
df_copy = df_copy.dropna(subset = ["bill_length_mm", "body_mass_g", "flipper_length_mm", "island", "species"])

# Use correct figure size
plt.figure(figsize=(15, 5))# Size

# (1) Scatter: bill_length_mm vs body_mass_g by species
plt.subplot(1, 3, 1)

species_colors = {
    'Adelie': 'blue',
    'Gentoo': 'red',
    'Chinstrap': 'green'
}
# Mask we need,directly use unique here
for dspecies in df_copy['species'].unique():
    mask = df_copy['species'] == dspecies
    subset = df_copy[mask]
    plt.scatter(
        subset['bill_length_mm'],
        subset['body_mass_g'],
        alpha=0.6,
        color=species_colors[dspecies],
        label=dspecies
    )

plt.xlabel('Bill Length (mm)')
plt.ylabel('Body Mass (g)')
plt.title('Bill Length vs Body Mass')
plt.legend()

# (2) Box plot: flipper_length_mm by island (raw arrays to plt.boxplot)
plt.subplot(1, 3, 2)

# Like .unique() in plot1,but will be used later
disland = df_copy['island'].unique()
# List of arrays for each island about required parameter
nparrays = []
for island in disland:
    mask = df_copy['island'] == island
    nparrays.append(df_copy[mask]['flipper_length_mm'].values)

plt.boxplot(nparrays, tick_labels = disland)
plt.xlabel('Island')
plt.ylabel('Flipper Length (mm)')
plt.title('Flipper Length by Island')

# (3) Histogram: body_mass_g
plt.subplot(1, 3, 3)

plt.hist(df_copy['body_mass_g'], bins=20, edgecolor='black')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.title('Body Mass Histogram')

# Layout, save
plt.tight_layout()
plt.savefig('penguin_viz.png')
plt.show()
````
